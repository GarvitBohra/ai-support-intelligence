```python

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# 1. Prepare the input text and target label
X = (
    df["Ticket Subject"].fillna("").astype(str)
    + " "
    + df["Ticket Description"].fillna("").astype(str)
)

y = df["Ticket Type"]

# Exclude rows with missing target labels
valid = y.notna()
X = X[valid]
y = y[valid]

# 2. Split data before training
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training examples:", len(X_train))
print("Testing examples:", len(X_test))

# 3. Build the ML pipeline
model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            lowercase=True,
            ngram_range=(1, 2),
            max_features=50000,
            sublinear_tf=True
        )
    ),
    (
        "classifier",
        LogisticRegression(
            max_iter=1000,
            random_state=42
        )
    )
])

# 4. Train the model
model.fit(X_train, y_train)

# 5. Evaluate on unseen test data
y_pred = model.predict(X_test)

print(f"\nAccuracy: {accuracy_score(y_test, y_pred):.4f}")
print("\nClassification report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python

from sklearn.base import clone
from sklearn.metrics import accuracy_score, f1_score
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# Keep exactly the same train/test split for a fair comparison
text_variants = {
    "Subject only": (
        df.loc[X_train.index, "Ticket Subject"].fillna("").astype(str),
        df.loc[X_test.index, "Ticket Subject"].fillna("").astype(str),
    ),
    "Description only": (
        df.loc[X_train.index, "Ticket Description"].fillna("").astype(str),
        df.loc[X_test.index, "Ticket Description"].fillna("").astype(str),
    ),
    "Subject + description": (X_train, X_test),
}

results = []

for name, (train_text, test_text) in text_variants.items():
    experiment = Pipeline([
        ("tfidf", TfidfVectorizer(
            lowercase=True,
            ngram_range=(1, 2),
            max_features=50000,
            sublinear_tf=True
        )),
        ("classifier", LogisticRegression(
            max_iter=1000,
            random_state=42
        ))
    ])

    experiment.fit(train_text, y_train)
    predictions = experiment.predict(test_text)

    results.append({
        "Input": name,
        "Accuracy": round(accuracy_score(y_test, predictions), 4),
        "Macro F1": round(
            f1_score(y_test, predictions, average="macro"),
            4
        )
    })

results_df = pd.DataFrame(results)
display(results_df)
```
